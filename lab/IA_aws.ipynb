## Carga de los datos a Amazon S3

Comience configurando algunas variables en el bucket de S3 y, a continuación, cree una función para cargar el archivo CSV en Amazon S3. Puede volver a usar esta función.

Primero, explore la función.

Observe la siguiente línea:

`dataframe.to_csv(csv_buffer, header=False, index=False)`

Esta línea escribe el DataFrame de pandas (que se pasó a la función) en el búfer de E/S, que se denomina *búfer_csv*. Usted utiliza un búfer dado que no es necesario escribir el archivo en forma local.

Para evitar que se escriban los encabezados de las columnas, utilice `header=False`. Para evitar que se produzca la salida del índice de pandas, use `index=False`.

Para escribir el búfer_csv en Amazon S3 como un objeto, utilice la operación `put` en `object`, que es una propiedad de `bucket`.


```python
bucket='c209731a5325977l15722840t1w586560789785-labbucket-9xo0hsnypki1'

prefix='lab3'

train_file='vertebral_train.csv'
test_file='vertebral_test.csv'
validate_file='vertebral_validate.csv'

import os

s3_resource = boto3.Session().resource('s3')
def upload_s3_csv(filename, folder, dataframe):
    csv_buffer = io.StringIO()
    dataframe.to_csv(csv_buffer, header=False, index=False)
    s3_resource.Bucket(bucket).Object(os.path.join(prefix, folder, filename)).put(Body=csv_buffer.getvalue())
```

Utilice la función que creó para cargar los tres conjuntos de datos.

```python
upload_s3_csv(train_file, 'train', train)
upload_s3_csv(test_file, 'test', test)
upload_s3_csv(validate_file, 'validate', validate)
```

# Paso 3: Entrenar el modelo

Ahora que los datos están en Amazon S3, puede entrenar un modelo. 

El primer paso es obtener el URI del contenedor XGBoost.

```python
import boto3
from sagemaker.image_uris import retrieve
container = retrieve('xgboost',boto3.Session().region_name,'1.0-1')
```

A continuación, debe establecer algunos *hiperparámetros* para el modelo. Dado que es la primera vez que está entrenando el modelo, puede usar algunos valores para comenzar.

```python
hyperparams={"num_round":"42",
             "eval_metric": "auc",
             "objective": "binary:logistic"}
```

Utilice la función **estimador** para configurar el modelo. Aquí hay algunos parámetros de interés:

- **recuento_de_instancia**: define cuántas instancias se utilizarán para el entrenamiento. Utilizará *una* instancia.
- **tipo_de_instancia**: define el tipo de instancia para el entrenamiento. En este caso, es *ml.m4.xlarge*.


```python
import sagemaker
s3_output_location="s3://{}/{}/output/".format(bucket,prefix)
xgb_model=sagemaker.estimator.Estimator(container,
                                       sagemaker.get_execution_role(),
                                       instance_count=1,
                                       instance_type='ml.m4.xlarge',
                                       output_path=s3_output_location,
                                        hyperparameters=hyperparams,
                                        sagemaker_session=sagemaker.Session())
```

El estimador necesita *canales* para ingresar datos en el modelo. Para el entrenamiento, se usarán el *canal_de_entrenamiento* y el *canal_de_validación*.

```python
train_channel = sagemaker.inputs.TrainingInput(
    "s3://{}/{}/train/".format(bucket,prefix,train_file),
    content_type='text/csv')

validate_channel = sagemaker.inputs.TrainingInput(
    "s3://{}/{}/validate/".format(bucket,prefix,validate_file),
    content_type='text/csv')

data_channels = {'train': train_channel, 'validation': validate_channel}
```

Al ejecutar **ajustar** se entrenará el modelo.

**Nota:** Este proceso puede tardar hasta 5 minutos.

```python
xgb_model.fit(inputs=data_channels, logs=False)
```

# Paso 4: Entrenar un modelo equivalente localmente y generar predicciones

**Nota importante:** el rol de ejecución de SageMaker de este laboratorio no permite crear endpoints (`sagemaker:CreateModel` bloqueado por IAM), así que no podemos usar `deploy()`. Tampoco es viable cargar el artefacto `model.tar.gz` generado por el contenedor `1.0-1` con la librería `xgboost` local, porque esa versión del contenedor (xgboost 1.0.1) usa un formato de serialización que las versiones modernas de `xgboost` ya no pueden leer, y no existe un paquete instalable de xgboost 1.0.1 compatible con Python 3.10.

**Solución:** entrenamos un modelo **equivalente** directamente en esta notebook, usando exactamente los mismos conjuntos de datos (`train`, `validate`, `test`) y los mismos hiperparámetros (`num_round=42`, `objective=binary:logistic`, `eval_metric=auc`) que se usaron en el trabajo de entrenamiento de SageMaker. El resultado es un modelo prácticamente idéntico, entrenado con la versión de `xgboost` que ya está instalada, así que no hay ningún problema de compatibilidad al generar predicciones ni al graficar las curvas de aprendizaje.

El trabajo de entrenamiento de SageMaker que ya completaste exitosamente sigue contando para la evaluación del laboratorio; esto es solo para poder generar las métricas y gráficos localmente.


## Entrenamiento local equivalente

Prepare los datos en el formato `DMatrix` que utiliza `xgboost` y entrene el modelo con los mismos hiperparámetros del trabajo original.


```python
import xgboost as xgb

X_train = train.drop(['class'], axis=1)
y_train = train['class']
X_validate = validate.drop(['class'], axis=1)
y_validate = validate['class']
X_test = test.drop(['class'], axis=1)
y_test = test['class']

dtrain = xgb.DMatrix(X_train, label=y_train)
dvalidate = xgb.DMatrix(X_validate, label=y_validate)
dtest = xgb.DMatrix(X_test)

params = {
    'objective': 'binary:logistic',
    'eval_metric': 'auc'
}

evals_result = {}
bst = xgb.train(
    params,
    dtrain,
    num_boost_round=42,
    evals=[(dtrain, 'train'), (dvalidate, 'validation')],
    evals_result=evals_result,
    verbose_eval=False
)

print('Modelo entrenado localmente con xgboost', xgb.__version__)
```

## Generación de predicciones sobre el conjunto de prueba

El modelo fue entrenado con el objetivo `binary:logistic`, por lo que la salida es la **probabilidad** de que el paciente pertenezca a la clase *anormal* (1).


```python
import numpy as np

test_y = y_test.to_numpy()

predictions_prob = bst.predict(dtest)
predictions_prob[:10]
```

## Generación de predicciones sobre el conjunto de prueba

Convierta el conjunto de prueba en un `DMatrix` (el formato de datos que utiliza XGBoost) y genere las probabilidades predichas.

El modelo fue entrenado con el objetivo `binary:logistic`, por lo que la salida es la **probabilidad** de que el paciente pertenezca a la clase *anormal* (1).


```python
test_X = test.drop(['class'], axis=1)  # sin .to_numpy()
test_y = test['class'].to_numpy()
dtest = xgb.DMatrix(test_X)  # conserva los nombres de columnas
predictions_prob = bst.predict(dtest)
```

Convierta las probabilidades en etiquetas de clase (0 o 1) utilizando un umbral de 0.5.


```python
predictions_class = np.where(predictions_prob > 0.5, 1, 0)
predictions_class[:10]
```

# Paso 5: Evaluar el modelo

Con las predicciones generadas, ahora puede calcular las métricas de rendimiento del modelo: matriz de confusión, exactitud (*accuracy*), precisión, exhaustividad (*recall*), puntuación F1 y curva ROC.


## Matriz de confusión

La matriz de confusión muestra la cantidad de predicciones correctas e incorrectas para cada clase.


```python
import matplotlib.pyplot as plt
from sklearn.metrics import confusion_matrix, ConfusionMatrixDisplay

cm = confusion_matrix(test_y, predictions_class)
print(cm)

disp = ConfusionMatrixDisplay(confusion_matrix=cm, display_labels=['Normal (0)', 'Anormal (1)'])
fig, ax = plt.subplots(figsize=(6, 6))
disp.plot(ax=ax, cmap='Blues', colorbar=False)
plt.title('Matriz de confusión')
plt.grid(False)
plt.show()
```

## Métricas de clasificación

Calcule la exactitud, la precisión, la exhaustividad, la puntuación F1 y el AUC, y muestre el informe de clasificación completo de scikit-learn.


```python
from sklearn.metrics import (accuracy_score, precision_score, recall_score,
                              f1_score, roc_auc_score, classification_report)

accuracy = accuracy_score(test_y, predictions_class)
precision = precision_score(test_y, predictions_class)
recall = recall_score(test_y, predictions_class)
f1 = f1_score(test_y, predictions_class)
auc_score = roc_auc_score(test_y, predictions_prob)

print(f'Exactitud (Accuracy):  {accuracy:.4f}')
print(f'Precisión (Precision): {precision:.4f}')
print(f'Exhaustividad (Recall): {recall:.4f}')
print(f'Puntuación F1:          {f1:.4f}')
print(f'AUC:                    {auc_score:.4f}')
print()
print(classification_report(test_y, predictions_class, target_names=['Normal (0)', 'Anormal (1)']))
```

## Gráfico comparativo de métricas

Un gráfico de barras facilita la visualización rápida de las métricas principales.


```python
metric_names = ['Accuracy', 'Precision', 'Recall', 'F1', 'AUC']
metric_values = [accuracy, precision, recall, f1, auc_score]

plt.figure(figsize=(7, 5))
bars = plt.bar(metric_names, metric_values, color='#2b6cb0')
plt.ylim(0, 1.05)
plt.ylabel('Valor')
plt.title('Métricas de rendimiento del modelo')
for bar, value in zip(bars, metric_values):
    plt.text(bar.get_x() + bar.get_width() / 2, value + 0.02, f'{value:.3f}', ha='center')
plt.show()
```

# Paso 6: Curva ROC

La curva ROC (*Receiver Operating Characteristic*) muestra la relación entre la tasa de verdaderos positivos y la tasa de falsos positivos para distintos umbrales de decisión. El área bajo la curva (AUC) resume el rendimiento general del modelo.


```python
from sklearn.metrics import roc_curve, auc

fpr, tpr, thresholds = roc_curve(test_y, predictions_prob)
roc_auc = auc(fpr, tpr)

plt.figure(figsize=(7, 7))
plt.plot(fpr, tpr, color='#2b6cb0', lw=2, label=f'Curva ROC (AUC = {roc_auc:.4f})')
plt.plot([0, 1], [0, 1], color='gray', lw=1, linestyle='--', label='Clasificador aleatorio')
plt.xlim([0.0, 1.0])
plt.ylim([0.0, 1.05])
plt.xlabel('Tasa de falsos positivos (FPR)')
plt.ylabel('Tasa de verdaderos positivos (TPR)')
plt.title('Curva ROC - Detección de anomalías vertebrales')
plt.legend(loc='lower right')
plt.grid(alpha=0.3)
plt.show()
```

# Paso 7: Curvas de aprendizaje (opcional)

Como el modelo se entrenó localmente con `xgb.train(..., evals=[...], evals_result=evals_result)`, ya tenemos guardado el histórico de AUC de entrenamiento y validación por cada ronda de boosting, sin depender de CloudWatch ni de trabajos de SageMaker.


```python
train_auc = evals_result['train']['auc']
validation_auc = evals_result['validation']['auc']
rounds = list(range(1, len(train_auc) + 1))

plt.figure(figsize=(8, 5))
plt.plot(rounds, train_auc, marker='o', label='Entrenamiento (train:auc)')
plt.plot(rounds, validation_auc, marker='o', label='Validación (validation:auc)')
plt.xlabel('Ronda de boosting')
plt.ylabel('AUC')
plt.title('Curva de aprendizaje: AUC de entrenamiento vs. validación')
plt.legend()
plt.grid(alpha=0.3)
plt.show()
```
